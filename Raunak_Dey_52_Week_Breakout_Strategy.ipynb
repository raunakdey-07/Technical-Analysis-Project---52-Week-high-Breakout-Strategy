```python
import yfinance as yf
import pandas as pd

def get_stock_data(stock_name):
    # Download stock data
    stock_data = yf.download(stock_name, period="1y", interval="1d")

    # Calculate rolling mean, rolling maximum, and EMAs
    stock_data['Rolling_Mean_52'] = stock_data['Close'].rolling(window=52).mean()
    stock_data['Rolling_Max_52'] = stock_data['High'].rolling(window=52).max()
    stock_data['EMA_5'] = stock_data['Close'].ewm(span=5, adjust=False).mean()
    stock_data['EMA_26'] = stock_data['Close'].ewm(span=26, adjust=False).mean()
    stock_data['SMA_Volume_20'] = stock_data['Volume'].rolling(window=20).mean()

    return stock_data

def main(stock_name):
    # Get stock data
    stock_data = get_stock_data(stock_name)

    # Apply conditions
    filtered_data = stock_data[
        (stock_data['Close'].shift(-52) > stock_data['Rolling_Mean_52']) &
        (stock_data['Close'] > stock_data['Close'].shift(1)) &
        (stock_data['EMA_5'] > stock_data['EMA_26']) &
        (stock_data['Volume'] > stock_data['SMA_Volume_20']) &
        (stock_data['High'] > stock_data['Rolling_Max_52'].shift(1))
    ]

    return filtered_data

if __name__ == "__main__":
    # Prompt user for stock name
    stock_name = input("Enter the stock name: ")

    # Call main function and print filtered data
    filtered_data = main(stock_name)
    print(filtered_data)
```

3. Inside the `get_stock_data` function, several columns are added to the `stock_data` DataFrame using the `rolling` and `ewm` functions from pandas. These columns are used to calculate the rolling mean, rolling maximum, and exponential moving averages (EMA) of the 'Close' column.

4. The `main` function is defined, which takes a `stock_name` parameter. It calls the `get_stock_data` function to retrieve the stock data and assigns it to the `stock_data` variable.

5. Inside the `main` function, several conditions are applied to the `stock_data` DataFrame using boolean indexing. These conditions are based on the criteria you provided in your question:
   - Weekly close should be above the previous 52-week closing
   - Daily close should be greater than the previous day's close
   - Daily EMA(Close, 5) should be greater than EMA(Close, 26)
   - Daily volume should be greater than the SMA volume(Close, 20)
   - Daily high should be greater than the previous 52-day high

6. The filtered data is returned from the `main` function.

7. In the main part of the script, the user is prompted to enter the name of the stock they want to analyze. The `main` function is called with the entered stock name, and the resulting filtered data is printed.
